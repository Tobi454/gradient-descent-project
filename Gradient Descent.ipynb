# Gradient Descent

In this project, I will implement gradient descent using Python, NumPy and Matplotlib to visualize how the algorithm approaches the minimum of a function.

The goal is to understand how derivatives and learning rates affect optimization.

```python
import numpy as np
import matplotlib.pyplot as plt
```

## 1. Defining the Function

I will start with the function:

f(x) = x²

The goal is to find the value of x that minimizes this function.

```python
#Defining the Function that calculates x²
def sqr(n):
    square = n*n
    return square
```

```python
print(sqr(-3))
print(sqr(2))
print(sqr(0))
```

I will plot the function f(x) = x² to understand its shape and identify its minimum.

```python
array = sqr(a)
```

## 3.  Derivative of the Function

The goal is to find the derivative of the function being considered.

```python
#Defining the Derivative function for x²
def derivative(n):
    derived = 2*n
    return derived
```

```python
print(derivative(-3))
print(derivative(2))
print(derivative(0))
```

## 4. Gradient Descent Update

I will use the derivative() function and the update rule to calculate the new value of x after just one update and Print the old value and the new value, assuming a learning rate of 1.

```python
#initial value
x = 4
```

```python
#learning rate
learning_rate = 0.1
```

```python
#new value after learning
x_new = x - learning_rate * (derivative(x))
print(x)
print(x_new)
```

```python
#difference in values inside the function(x²) before and after learning 
dif = sqr(x) - sqr(x_new)
print(dif)
```

```python
#performing the gradient update ten times
x = 4
i = 0
while(i<10):
    x_new = x - learning_rate * derivative(x)
    print("X_Old:",x,"-->","X_New",x_new)
    x = x_new
    i = i + 1
```

```python
#storing the values in an array
x_list = []
x_squared_list = []
x = 4
i = 0
while(i<10):
    x_new = x - learning_rate * derivative(x)
    x_list.append(x)
    x_squared_list.append(sqr(x))
    x = x_new
    i = i + 1
```

```python
#values after gradient updates
print(x_list)
```

```python
#values of the several x in the function
print(x_squared_list)
```

Using different learning rates while keeping other values constant

```python
#Using a learning rate of 0.01
x_list = []
x_squared_list = []
x = 4
i = 0
while(i<10):
    #x_new = x - learning_rate * derivative(x)
    x_new = x - 0.01 * derivative(x)
    x_list.append(x)
    x_squared_list.append(sqr(x))
    x = x_new
    i = i + 1

print(x_list[9])
print(x_squared_list[9]) 



plt.plot(a,array)
plt.xlabel("Number")
plt.ylabel("Squared Value")
plt.title("Graph of f(x) = x²")
plt.scatter(x=x_list,y=x_squared_list)
plt.grid()

plt.show()
```

```python
#Using a learning rate of 0.1
x_list = []
x_squared_list = []
x = 4
i = 0
while(i<10):
    #x_new = x - learning_rate * derivative(x)
    x_new = x - 0.1 * derivative(x)
    x_list.append(x)
    x_squared_list.append(sqr(x))
    x = x_new
    i = i + 1

print(x_list[9])
print(x_squared_list[9]) 


plt.plot(a,array)
plt.xlabel("Number")
plt.ylabel("Squared Value")
plt.title("Graph of f(x) = x²")
plt.scatter(x=x_list,y=x_squared_list)
plt.grid()

plt.show()
```

```python
#Using a learning rate of 0.5
x_list = []
x_squared_list = []
x = 4
i = 0
while(i<10):
    #x_new = x - learning_rate * derivative(x)
    x_new = x - 0.5 * derivative(x)
    x_list.append(x)
    x_squared_list.append(sqr(x))
    x = x_new
    i = i + 1

print(x_list[9])
print(x_squared_list[9])


plt.plot(a,array)
plt.xlabel("Number")
plt.ylabel("Squared Value")
plt.title("Graph of f(x) = x²")
plt.scatter(x=x_list,y=x_squared_list)
plt.grid()

plt.show()
```

```python
#Using a large learning rate of 1.1
x_list = []
x_squared_list = []
x = 4
i = 0
while(i<10):
    #x_new = x - learning_rate * derivative(x)
    x_new = x - 1.1 * derivative(x)
    x_list.append(x)
    x_squared_list.append(sqr(x))
    x = x_new
    i = i + 1

print(x_list[9])
print(x_squared_list[9])


plt.plot(a,array)
plt.xlabel("Number")
plt.ylabel("Squared Value")
plt.title("Graph of f(x) = x²")
plt.scatter(x=x_list,y=x_squared_list)
plt.grid()

plt.show()
```

Observations
1. Learning rate of 0.01: The algorithm made very small updates. After 10 iterations, it was still far from the minimum, showing that a very small learning rate can make convergence slow.

2. Learning rate of 0.1: The algorithm made larger updates and moved much closer to the minimum within the same number of iterations.

3. Learning rate of 0.5: The algorithm reached the minimum at x = 0. For this particular function, this learning rate was sufficient to reach the minimum in a single update.

4. Learning rate of 1.1: The algorithm overshot the minimum repeatedly, moving between positive and negative values while getting farther away from zero. This resulted in divergence.
Conclusion

The experiments show that learning rate plays an important role in the performance of gradient descent. A learning rate that is too small can make convergence slow, while one that is too large can cause the algorithm to overshoot the minimum and diverge.

A suitable learning rate allows gradient descent to approach the minimum efficiently without taking excessively large steps.

## 6. Customizable Gradient Descent

Making the Gradient Descent Algorithm Flexible

```python
#x - starting value, rate - learning rate, n - number of iterations
def gradient(x,rate,n):
    x_list = []
    x_squared_list = []
    i = 0
    while(i<n):
        x_new = x - rate * derivative(x)
        x_list.append(x)
        x_squared_list.append(sqr(x))
        x = x_new
        i = i + 1

    return x, x_list, x_squared_list
```

```python
#testing the function using x - 4, rate - 0.1, n - 10
print(gradient(4,0.1,10))
```

```python
#testing the function using x - 6, rate - 0.1, n - 10
print(gradient(6,0.1,10))
```

```python
#testing the function using x - 4, rate - 0.01, n - 20
print(gradient(4,0.01,20))
```

## 7. Customizable Function

## 8. Adding a tolerance range for Convergence



```python
#using f(x) = x² + 4 + 5
def f(x):
    return x**2 + 4*x + 5


def df(x):
    return 2*x + 4


#x - starting value, rate - learning rate, n - number of iterations, f - function, df - derivative, tol -  tolerance
def gradient(x, rate, n, f, df, tol):

    x_list = []
    y_list = []
    i = 0

    while i < n:
        x_new = x - rate * df(x)

        x_list.append(x)
        y_list.append(f(x))

        x = x_new
        i = i + 1

        if abs(df(x)) < tol:
            break

    return x, x_list, y_list, i
```

```python
gradient(2, 0.1, 10, f, df, 0.5)
```
